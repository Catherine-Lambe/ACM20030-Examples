# Question 1: Ordinary Differential Equations (ODEs)

All the below questions should be answered by using `RK2()` function below:

```python
# Arguments for the function:
# f - the right-hand side of dy/dx = f(x,y)
# dx - the step size Delta x
# x0 - the initial x value
# y0 - the initial y value(s)
# imax - the maximum number of steps to take
#
# The function returns an array containing the steps taken
def RK2(f, dx, x0, y0, imax):
    xi = x0
    yi = y0
    
    # Create arrays to store the steps in    
    xis = np.zeros((imax+1,1))
    yis = np.zeros((imax+1,np.size(y0)))
    xis[0] = x0
    yis[0] = y0
    
    i = 0
    while i < imax:
        k1 = dx*f(xi, yi)
        k2 = dx*f(xi + dx, yi + k1)
    
        yi = yi + 0.5*(k1 + k2)
        
        xi += dx
        i += 1
        
        # Store the steps
        xis[i] = xi
        yis[i] = yi  
        
    return [xis,yis]
```

# Question 2

Use the `SimpsonsIntegrate` function below when answering this question

```python
def SimpsonsIntegrate(f, a, b, N):
    dx = (b-a)/N
    xi = a
    i = 0
    area = 0
    while i < N:
        area += dx/6*(f(xi) + 4*f((2*xi+dx)/2) + f(xi+dx))
        xi += dx
        i+= 1
        
    return area
```

# Question 3: fitting data

Use the below `LinearLeastSquares` function to answer the below questions

```python
# Function arguments:
# xi - the x data
# yi - the y data
# The function returns x where x[0] = m and x[1] = c where the model is y = mx + c
def LinearLeastSquares(xi, yi):
    # Compute the 5 coefficients in the equations
    n = xi.size
    sum_xi2 = np.sum(xi**2)
    sum_xi  = np.sum(xi)
    sum_yi  = np.sum(yi)
    sum_xiyi= np.sum(xi*yi)
    # Write the normal equations as a matrix equation A.x = b
    # where x = {m c}, b = {xiyi, yi} and A = {{xi2, xi}, {xi,n}}
    # Then x = A^{-1}.b
    Ainv = 1/(sum_xi2*n - sum_xi**2)*np.array([[n, -sum_xi], [-sum_xi, sum_xi2]])
    b = np.array([sum_xiyi, sum_yi])
    x = np.dot(Ainv,b)
    #  x[0] = m,  x[1] = c
    return x
```
